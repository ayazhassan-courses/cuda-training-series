# CUDA Training Resource

The materials in this repository accompany the CUDA Training Series presented at ORNL and NERSC.

You can find the slides and presentation recordings at https://www.olcf.ornl.gov/cuda-training-series/

## Repository Overview

This repository contains CUDA training materials organized into homework exercises covering various aspects of CUDA programming and GPU computing.

## Getting Started

To get started with these CUDA training materials:

1. **Prerequisites**: Basic knowledge of C/C++ programming
2. **Hardware Requirements**: Access to a CUDA-capable GPU
3. **Software Requirements**: CUDA Toolkit installed
4. **Training Series**: Follow along with the presentation materials available at the link above

## Exercises Structure

The exercises are organized in the `exercises/` directory with homework assignments numbered from `hw1` to `hw13`. Each homework covers specific CUDA programming concepts and techniques.
